```python
data  ={
    'Date': pd.date_range('2023-01-01',periods=20),
    'Product': ['A','B','C','D']*5,
    'Region': ['East', 'West', 'North', 'South', 'South', 'East', 'West', 'North', 'East', 'South', 'West', 'North', 'North', 'East', 'South', 'West', 'East', 'North', 'South', 'West'],
    'Sales': np.random.randint(100,1000,20),
    'Units':np.random.randint(10,50,20),
    'Rep': ['Bob', 'Mary', 'Alice', 'John', 'John', 'Bob', 'Mary', 'Alice', 'Bob', 'John', 'Mary', 'Alice', 'Alice', 'John', 'Bob', 'Mary', 'Bob', 'Alice', 'Mary', 'John']

}
df = pd.DataFrame(data)
df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
```

```python
df
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='sum')
```

```python
#cross Tabs
```

```python
pd.crosstab(df['Region'],df['Product'])
```
